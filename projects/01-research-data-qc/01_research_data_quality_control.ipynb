```python

import pandas as pd

# Simulated research data (not real patient data)
data = {
    "participant": ["P001", "P001", "P001",
                    "P002", "P002", "P002"],
    "timepoint": ["Baseline", "60min", "120min",
                  "Baseline", "60min", "120min"],
    "expected_min": [0, 60, 120, 0, 60, 120],
    "actual_min": [0, 64, 119, 0, 79, None]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Flag records with missing collection timestamps
df["missing_collection_time"] = df["actual_min"].isna()

# Calculate collection timing differences
df["time_difference"] = (
    df["actual_min"] - df["expected_min"]
).abs()

# Flag samples outside a 10-minute window
df["outside_window"] = df["time_difference"] > 10

# Create an overall QC status
df["qc_status"] = "PASS"

df.loc[df["outside_window"], "qc_status"] = "LATE_OR_EARLY"
df.loc[df["missing_collection_time"], "qc_status"] = "MISSING_TIME"

print(df[["participant", "timepoint", "qc_status"]])
```

```python
df["qc_status"].value_counts()
```

```python
print(df.columns)
```

```python
proportion_missing = df["missing_collection_time"].sum() / len(df)
percentage_missing = proportion_missing * 100
rounded_percentage = round(percentage_missing, 2)
print(rounded_percentage)
```

## Day 2 — MRI Quality Control Metrics

### Learning Objectives
- Count missing records using `.count()`
- Calculate dataset size using `len()`
- Calculate percentages
- Round numeric results using `round()`
- Format research summaries with Python f-strings

```python
# Day 2: MRI Research Quality Control

# Example dataset
scan_status = [
    "complete",
    "complete",
    "missing",
    "complete",
    "incomplete",
    "complete"
]

# Count missing scans
missing_count = scan_status.count("missing")

# Calculate total scans
total_scans = len(scan_status)

# Calculate missing percentage
missing_percentage = (missing_count / total_scans) * 100

# Round to two decimal places
rounded_percentage = round(missing_percentage, 2)

# Print QC summary
print(f"Missing MRI scans: {missing_count} out of {total_scans} ({rounded_percentage}%)")
```

## Day 3 — Loops and Conditional QC Checks

```python
scan_status = [
    "complete",
    "missing",
    "complete",
    "unknown",
    "incomplete"
]

for scan in scan_status:
    if scan == "missing":
        print("Missing MRI scan detected")
    elif scan == "incomplete":
        print("Incomplete MRI scan detected")
    elif scan == "complete":
        print("MRI scan complete")
    else:
        print("Unknown scan status - review required")
```
